## 6. Interpretation

- **Strong relationship with Total Sales:** Rating shows the strongest positive correlation with Total Sales (r ≈ 0.57), followed by Quantity (r ≈ 0.34). Unit Price shows only a weak positive relationship (r ≈ 0.26).
- **Best-performing product line:** Electronics has the highest average Total Sales, closely followed by Fashion Accessories.
- **Customer type contribution:** Member customers contribute far more to total sales (₹5,312.50) than Normal customers (₹3,790.90).
- **Rating vs sales relationship:** Yes, higher customer ratings correspond to higher total sales (r ≈ 0.57).

### Five Key Business Insights
1. Member customers consistently outspend Normal customers across every city — grow the loyalty program.
2. Electronics and Fashion Accessories are the top two revenue-generating product lines — prioritize stock and marketing.
3. Higher customer ratings correlate with higher sales — improving customer experience can boost revenue.
4. Female customers show a higher average spend than male customers — consider gender-targeted marketing.
5. Payment method usage is evenly split across Cash, Credit Card, and E-wallet — continue supporting all channels.
